# PVFS to EDF conversion example
This notebook runs one conversion command from a source PVFS file to an EDF output file.

```python
from pathlib import Path
import shlex
import subprocess

# Update these paths for your machine
source_file = Path('/Volumes/eeg-lab/recorded-pvfs/example_recording.pvfs')
sink_dir = Path('/Volumes/eeg-results/edf')

# Vendor converter command template
command_template = 'pvfs2edf --input {input} --output {output}'

sink_dir.mkdir(parents=True, exist_ok=True)
output_file = sink_dir / (source_file.stem + '.edf')

cmd = command_template.format(
    input=shlex.quote(str(source_file)),
    output=shlex.quote(str(output_file)),
)

print('Running:', cmd)
result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
print('Return code:', result.returncode)
if result.stdout.strip():
    print('STDOUT:\n' + result.stdout)
if result.stderr.strip():
    print('STDERR:\n' + result.stderr)
```

```python
print('Output exists:', output_file.exists())
print('Output path:', output_file)
```
